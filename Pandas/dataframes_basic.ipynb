```python
person = {
    "first": "Corey", 
    "last": "Schafer", 
    "email": "CoreyMSchafer@gmail.com"
}
```

```python
people = {
    "first": ["Corey"], 
    "last": ["Schafer"], 
    "email": ["CoreyMSchafer@gmail.com"]
}
```

```python
people = {
    "first": ["Corey", 'Jane', 'John'], 
    "last": ["Schafer", 'Doe', 'Doe'], 
    "email": ["CoreyMSchafer@gmail.com", 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
people['email']
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people) # passing the dictionary that has the values as the list
```

```python
df # represents in a way of rows and columns which we can visualise
# 0,1,2 -> unique value for our rows
```

```python
df['email']
# the output of this doesn't really match to when we returned it from the dictionary
# when we printed from the dictionary -> output was in the form of a list
# here it is in form of a series
```

```python
type(df['email']) # -> series object
```

```python
# series is still basically a list of data but it corresponds to a dataframe
# kindof like a 1D array
# it's basically rows of data
```

```python
# dataframe -> rows and columns
# series -> rows of a single column

# hence dataframe is a container for multiple series objects
```

```python
# dot notation
df.email
```

```python
# accessing mutiple columns
# pass a list inside the brackets
df[['last', 'email']]
# here we get multiple columns as output hence it's not longer a series
# now it is just a filtered down dataframe 
```

```python
df.columns
# gives us all of the columns
```

```python
# to get rows
# iloc -> access rows by integer notation
df.iloc[0] # first row
```

```python
# can access multiple rows by passing a list of integers
df.iloc[[0,1]]  
```

```python
# with these iloc and loc we can pass columns as well
# 1st argument - rows
# 2nd argument - columns
df.iloc[[0, 1] , 2] # email adresses of first 2 rows i.e. 2nd column of 0th and 1st row
```

```python
# iloc -> we were searching by integer location
# loc -> will be searching by label
# labels for rows -> indexes 
```

```python
print(df)
```

```python
df.loc[0] 
# get the row with label 0
# looks similar to iloc but will click when we actually use it in a dataset
```

```python
# can also pass in a list to specify mutiple rows
df.loc[[0,1]]
# output is a datafram now that we passed multiple rows
```

```python
# in iloc we used inetegrs, in loc we can use labels
df.loc[[0,1], 'email']
# want email values only of the first 2 rows
```

```python
# if want email as well as last name, instead of passing just an email string, we can pass a list of strings
df.loc[[0,1], ['email', 'last']]
# the order in which we asked for is the order in which it is returned, we asked for email then last, it returned email then last
```
