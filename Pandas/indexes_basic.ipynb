```python
import pandas as pd
```

```python
people = {
    "first": ["Corey", 'Jane', 'John'], 
    "last": ["Schafer", 'Doe', 'Doe'], 
    "email": ["CoreyMSchafer@gmail.com", 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
# on the left we see a range of numbers as an identifier for each row
```

```python
# wanna view just the email addresses
df['email']
```

```python
# want to se email addresses as the indexes for the dataframes
df.set_index('email')
# can see now that emails are at the left and are also bold
```

```python
df
# here it did not change, it still has the default indexes
# this si because pandas doesn't do all of these changes in place unless we specifically tell it to do
```

```python
df.set_index('email', inplace=True)
# now the changes will reflect even in the dataframe
```

```python
df
```

```python
df.index
```

```python
# why is this useful?
# the email addresses give a nice unique identifier
# previously while using .loc, we had to search from indexes (0,1,2,3...), which is tedious and hard to read
# having a unique identifier is always wayyy better
# now we can find a row by passing it's specifi label and it will be much more easy
```

```python
df.loc['JaneDoe@email.com']
```

```python
# can also get specific values
df.loc['JaneDoe@email.com', 'last']
# df.loc[0] -> will give us an error, if we want to access using integer locations, use iloc
```

```python
df.iloc[0]
```

```python
# to reset index
df.reset_index(inplace = True)
df
```
