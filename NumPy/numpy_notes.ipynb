import numpy library

```python
import numpy as np
```

the basics

```python
a = np.array([1,2,3]) # 1D array
print(a)
```

```python
b = np.array([[9.0,8.0,7.0], [6.0,5.0,4.0]]) # nest list within a list
print(b)
```

```python
# getting dimension of the numpy array
a.ndim
```

```python
# getting dimension of the numpy array
b.ndim
```

```python
# get shape
a.shape
# a is 1D -> 3 rows, 1 column
# mistake here - in 1D array it doesn't actually have rows and columns at all, so output (3,) just means 3 elements in 1D 
```

```python
b.shape
# 2 rows, 3 columns
```

```python
# get type
# type of data stored
a.dtype 
# if wanted to specify what type we wanted it to store as -> a = np.array([1,2,3], dtype = 'int16')
```

```python
# get size
# tells how many bytes each element makes up (64 bits = 8 bytes)
a.itemsize
```

```python
# get total size
# total number of elemnts
a.size
```

```python
# total memory array takes up in bytes
a.size * a.itemsize
```

```python
# another way to do that
a.nbytes
```
