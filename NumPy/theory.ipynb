#### What is NumPy??
- Multidimensional array library
- Can store data in 1D / 2D / 3D arrays

#### Why use NumPy over Lists??

##### Performance Benefits
- Lists are slow, NumPy is very fast
- 5 -> Binary -> 00000101 (8 bit -> 1 byte) -> NumPy by default casts this into a 32 bit(int32), i.e. a 4 byte memory space(in NumPy we can specify if we want int8 or int16 according to our needs)
- Whereas in Lists, lists use a built in int type for python which consists of 4 different things - object value(represented as a long - 8 bytes), object type(8 bytes), reference count(8 bytes), size(4 bytes) -> hence we can see that a single integer ina  list requires a lot more space than a NumPy
- Hence as NumPy uses less bytes of memory it is faster to read
- When iterating each item in a numpy array, don't need to do typechecking each time
- NumPy uutilizes contiguous memory
- In lists - if we have a 8 valued array, the 8 values are just memory addresses of stuff which is in practicality scattered around and not stored in continuos blocks of memory, and all the info is not right next to each other and have to bounce around the computer's memory for everything, hence not super super fast
- NumPy uses contiguous memroy - all 8 blocks -> next to each other
- This leads to efficient utilisation of SIMD Vector Processing
- Also helps effectively utilize cache
